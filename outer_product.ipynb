```python
v = np.arange(1,11)
w = np.arange(1,6)
print(np.outer(v,w))
print(np.outer(w,v))
```

```python
s = 4
res1 = s * np.outer(v,w)
res2 = np.outer(s*v,w)
res3 = np.outer(v,s*w)
res4 = np.outer(v,w)*s
print(res1 - res2)
print(res1 - res3)
print(res1 - res4)
```
