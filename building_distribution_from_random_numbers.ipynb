```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
```

```python
shift = 5
stretch = 2
n = 1000
pnts = np.random.randn(n)*stretch + shift - .5*stretch
pnts = np.random.rand(n)*stretch + shift

print('Mean is %g, std is %g' %(np.mean(pnts), np.std(pnts)))
print('Mean is %g, range is %g' %(np.mean(pnts), np.max(pnts) - np.min(pnts)))
```

```python
fig ,ax = plt.subplots(1,2, figsize=(6,3))
ax[0].plot(pnts, 's', alpha=.4)
# ax[1].hist(pnts,bins=[4,4.1,4.5,5.4,5.5,6],edgecolor='k')
ax[1].hist(pnts,edgecolor='k')

plt.show()
```

```python
lam = 3.4
n = 1000
pnts = np.random.poisson(lam ,n)
print('Mean is %g, variance is %g'%(np.mean(pnts), np.var(pnts)))
fig ,ax = plt.subplots(1,2, figsize=(6,3))
ax[0].plot(pnts, 's', alpha=.4)
ax[1].hist(pnts,bins=np.arange(0,np.max(pnts)+1),edgecolor='w')

plt.show()
```

```python
from scipy.stats import rankdata
```

```python
n = 500
orig_data = 2**np.random.randn(n)


rank_data = rankdata(orig_data)
rank_data = np.append(rank_data, np.max(rank_data) + 1)

normal_data = 2*(rank_data/np.max(rank_data))-1
normal_data =np.delete(normal_data,n)

# apply the inverse hyperbolic transform

trans_data = np.arctanh(normal_data)

fig, ax = plt.subplots(1,2)

ax[0].hist(orig_data, 30)
ax[0].set_xlabel('Value')
ax[0].set_ylabel('Count')

ax[1].hist(trans_data, 30)
ax[1].set_xlabel('Value')
ax[1].set_ylabel('Count')
```

```python
n = 2000

# plt.plot(np.random.rand(n),'.')
# plt.plot(np.linspace(0.6*np.pi,n))
a = np.linspace(0,6*np.pi,n)
sig1 = np.sin(a)
# print(sig1.shape)
sig2 = np.random.rand(n)*2-1
# print(sig2.shape)

combi = sig1 + sig2

fig, ax = plt.subplots(2,3,figsize=(10,4))

ax[0,0].plot(sig1)
ax[0,0].set_title('Sine wave')
ax[1,0].hist(sig1,100)

ax[0,1].plot(sig2,'.',alpha=.3)
ax[0,1].set_title('Uniform numbers')
ax[1,1].hist(sig2,100)

ax[0,2].plot(combi,'.',alpha=.3)
ax[0,2].set_title('Combined signal')
ax[1,2].hist(combi,100)


for i in np.matrix.flatten(ax):
    i.set_xticks([])
    i.set_yticks([])
plt.show()
```

```python
ns = np.arange(50,10000,100)
nbins = 50
all_hists = np.zeros((len(ns),nbins))
for i in range(len(ns)):
    sig1 = np.sin(np.linspace(0,6*np.pi,ns[i]))
    sig2 = np.random.rand(ns[i])*2-1
    all_hists[i,:],x = np.histogram(sig1+sig2,nbins,density=True)

# plt.pcolor(x[:-1], ns, all_hists,vmax=.6,cmap='viridis')
plt.pcolor(x[:-1], ns, all_hists,vmax=.6,cmap='Purples')

plt.xlabel('Data value')
plt.ylabel('Points in the sample')
plt.colorbar()
plt.show()
```
