```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
```

```python
shift = 5
stretch = 2
n = 1000
pnts = np.random.randn(n)*stretch + shift - .5*stretch
pnts = np.random.rand(n)*stretch + shift

print('Mean is %g, std is %g' %(np.mean(pnts), np.std(pnts)))
print('Mean is %g, range is %g' %(np.mean(pnts), np.max(pnts) - np.min(pnts)))
```

```python
fig ,ax = plt.subplots(1,2, figsize=(6,3))
ax[0].plot(pnts, 's', alpha=.4)
# ax[1].hist(pnts,bins=[4,4.1,4.5,5.4,5.5,6],edgecolor='k')
ax[1].hist(pnts,edgecolor='k')

plt.show()
```

```python
lam = 3.4
n = 1000
pnts = np.random.poisson(lam ,n)
print('Mean is %g, variance is %g'%(np.mean(pnts), np.var(pnts)))
fig ,ax = plt.subplots(1,2, figsize=(6,3))
ax[0].plot(pnts, 's', alpha=.4)
ax[1].hist(pnts,bins=np.arange(0,np.max(pnts)+1),edgecolor='w')

plt.show()
```

```python
from scipy.stats import rankdata
```

```python
n = 500
orig_data = 2**np.random.randn(n)


rank_data = rankdata(orig_data)
rank_data = np.append(rank_data, np.max(rank_data) + 1)

normal_data = 2*(rank_data/np.max(rank_data))-1
normal_data =np.delete(normal_data,n)

# apply the inverse hyperbolic transform

trans_data = np.arctanh(normal_data)

fig, ax = plt.subplots(1,2)

ax[0].hist(orig_data, 30)
ax[0].set_xlabel('Value')
ax[0].set_ylabel('Count')

ax[1].hist(trans_data, 30)
ax[1].set_xlabel('Value')
ax[1].set_ylabel('Count')
```
