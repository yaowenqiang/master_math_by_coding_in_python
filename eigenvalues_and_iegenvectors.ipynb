```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
M = np.random.randint(-5,6,(5,5))
M = M@M.T 
# print(N)
eigvals,  eigvecs = np.linalg.eig(M)
print(eigvecs)
```

```python
fig, ax = plt.subplots(1,3)
ax[0].imshow(M)
ax[0].set_title('The matrix')
print(np.diag(eigvals))
ax[1].imshow(np.diag(eigvals))
ax[1].set_title('Eigenvalues')
ax[2].imshow(eigvecs)
ax[2].set_title('Eigenvectors')
```

```python
Mv = M@eigvecs[:,0]
lv = eigvals[0] * eigvecs[:,0]
print(Mv)
print(lv)
```

```python
M = np.array([[1,2],[2,1]])
d,v = np.linalg.eig(M)

lv = d[0]*v[:,0]
Mv = M@v[:,0]

lv = d[1]*v[:,1]
Mv = M@v[:,1]

# print(Mv)
plt.plot([0,lv[0]],[0,lv[1]],'b',label='$\\lambda v$',linewidth=4)
plt.plot([0,Mv[0]],[0,Mv[1]],'r:',label='Mv',linewidth=4)
plt.legend()
plt.grid()
plt.axis('square')
plt.axis([-3,3,-3,3])
plt.show()
```
