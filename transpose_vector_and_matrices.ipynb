```python
mat = np.random.randn(8,4)
mat_t = mat.T
fig, ax = plt.subplots(1,2)
ax[0].imshow(mat)
ax[0].set_title('M')
# ax[0].set_xticks([])
# ax[1].set_xticks([])
for i in ax:
    i.set_xticks([])
    i.set_yticks([])
ax[1].imshow(mat_t)
ax[1].set_title('$M^T$')
```
