```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
import seaborn as sns
```

```python
number_range = np.arange(2,51)
a = []
b = []
c = []
for m in number_range:
    for n in range(number_range[0],m): # m > n
        a.append(m**2 - n**2)
        b.append(2*m*n)
        c.append(m**2 + n**2)
        
    
```

```python
fig,ax = plt.subplots(1,figsize=(10,7))
plt.scatter(a,b,c=c,marker='o',alpha=.5,linewidths=.5,edgecolors='k')
plt.axis('off')
plt.show()
```

```python
pythcheck = np.array(a)**2 + np.array(b)**2 - np.array(c)**2
```
