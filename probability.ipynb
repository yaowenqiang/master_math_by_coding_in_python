```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
data = [1,2,2,2,2,2,3,5,6,6,6,7,8,9,8,6,8,3]
plt.hist(data)
plt.xlabel('Value')
plt.ylabel('Number of apperances (count)')
plt.show()

hist_data = np.histogram(data,bins=np.arange(1,11))
print(hist_data)

plt.bar(hist_data[1][:-1], hist_data[0])
plt.xlabel('Value')
plt.ylabel('Number of apperances (count)')
plt.show()
```

```python
y,x = np.histogram(data,bins=np.arange(1,11))

yP = y/(np.sum(y))
plt.bar(x[:-1], yP)

plt.xlabel('Value')
plt.ylabel('Proportion')
plt.xticks(np.arange(1,10))
plt.show()
```

```python
y,x = np.histogram(data,bins=np.arange(1,11))

yP = y/(np.max(y))
plt.bar(x[:-1], yP)

plt.xlabel('Value')
plt.ylabel('Proportion(norm.)')
plt.xticks(np.arange(1,10))
plt.show()
```

```python
x = np.linspace(-3,3,1000)
fig,ax = plt.subplots(2,3)
fig.set_size_inches(10,6)
for i in range(3):
    y = x**(i+1)
    h = np.histogram(y, 50)
    yP = h[0]/sum(h[0])
    ax[0,i].plot(x,y,'s')
    ax[0,i].set_title('$y=x^{%g}$'%(i+1))
    ax[1,i].plot(h[1][:-1],yP)
plt.show()
```
