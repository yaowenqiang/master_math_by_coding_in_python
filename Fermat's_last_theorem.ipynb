```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
import seaborn as sns
```

```python
ints = np.arange(1,101)
Z = np.zeros((len(ints),len(ints)))
for x in ints:
    for y in ints:
        Z[x-1,y-1] = (x**3 + y **3)**(1/3)

```

```python
z_ints = Z % 1  == 0
```

```python
z_ints
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,6))
ax[0].imshow(Z)
ax[0].set_xlabel('y')
ax[0].set_ylabel('x')
ax[0].set_title('$z = \sqrt[3]{x^3+y^3}$',fontsize=18)
ax[1].imshow(z_ints)
ax[1].set_xlabel('y')
ax[1].set_ylabel('x')
ax[1].set_title('Wether $z$ is an int',fontsize=18)

plt.show()
```

```python
ints = np.arange(1,101)
Z = np.zeros((len(ints),len(ints)))
for x in ints:
    for y in ints:
        Z[x-1,y-1] = (x**2 + y **2)**(1/2)

fig, ax = plt.subplots(1,2,figsize=(12,6))
ax[0].imshow(Z)
z_ints = Z % 1  == 0
ax[0].set_xlabel('y')
ax[0].set_ylabel('x')
ax[0].set_title('$z = \sqrt[2]{x^2+y^2}$',fontsize=18)
ax[1].imshow(z_ints)
ax[1].set_xlabel('y')
ax[1].set_ylabel('x')
ax[1].set_title('Wether $z$ is an int',fontsize=18)

plt.show()
```
