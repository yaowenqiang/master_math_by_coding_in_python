```python
nums_to_sqrt = np.linspace(2,101,50)
n_iterattions = np.arange(3,9)
err = np.zeros((len(n_iterattions),len(nums_to_sqrt)))

for ni, num in enumerate(nums_to_sqrt):
    Y = num
    for ii, iters in enumerate(n_iterattions):
        
        x = Y/3
        for n in range(iters):
            x = (x+Y/x)/2
        
        err[ii,ni] = abs(x-np.sqrt(Y))
        
```

```python
plt.imshow(-np.log(err),aspect=10,extent=[nums_to_sqrt[0],nums_to_sqrt[-1],n_iterattions[-1],n_iterattions[0]])
```

```python
num_2_sqrt = 13
start_est = np.linspace(-num_2_sqrt,num_2_sqrt,40)
n_iters = 5
sqrt_alg_results = np.zeros((len(start_est),n_iters+1))
for idx, start_num in enumerate(start_est):
    # sqrt_alg_results[idx,0] = start_num
    x = [start_num]
    # x = num_2_sqrt/3
    for n in range(n_iters):
        new_x = (x[n] + num_2_sqrt/x[n])/2
        x.append(new_x)
                
    sqrt_alg_results[idx,:] = x
```

```python
fig,ax = plt.subplots(1,figsize=(8,6))
plt.plot(np.arange(n_iters+1),sqrt_alg_results.T)
plt.show()
```
