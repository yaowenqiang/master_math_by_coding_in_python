```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
x = sym.symbols('x')
f = x
sym.integrate(f)
```

```python
sym.integrate(f,(x,0,1))
```

```python
p = sym.plotting.plot(f,show=False)
p.xlim = [0,1]
p.ylim = [0,1]
p.show()
```

```python
f = x**3/(x-2)
intf = sym.integrate(f)
p = sym.plotting.plot(f,show=False)
p.extend(sym.plotting.plot(intf, (x,2.1,10),show=False, line_color='r'))
p[0].label = '$f(x) = %s$'%(sym.latex(f))
p[1].label = '$\\int f(x) = %s$'%(sym.latex(intf))
p.legend = True
p.ylim=[-200,200]
p.show()
```

```python
x = sym.symbols('x')
f = 2*x**3 + sym.sin(x)
df = sym.diff(f)
idf = sym.integrate(df)

display(Math('f(x) = %s' %(sym.latex(f))))
display(Math('f\' = %s' %(sym.latex(df))))
display(Math('\\int (f\')dx = %s' %(sym.latex(idf))))
```
