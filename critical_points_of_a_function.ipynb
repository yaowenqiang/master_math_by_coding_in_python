```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
from scipy.signal import find_peaks
```

```python
# empirical method
x = np.linspace(-5,5,10001)
fx = x**2 * np.exp(-x**2)

dfx = np.diff(fx)/(x[1]-x[0])

local_max = find_peaks(fx)[0]
local_min = find_peaks(-fx)[0]

print(f'The critical points are {x[local_max]} {x[local_min]}')
plt.plot(x,fx)
plt.plot(x[local_max],fx[local_max],'ro')
plt.plot(x[local_min],fx[local_min],'gs')
plt.plot(x[0:-1],dfx)
plt.show()
```

```python
a,x = sym.symbols('a,x')
base_expr = x**2*sym.exp(-a*x**2)
arange = np.arange(0,2.25,.25)
xrange = np.linspace(-3,3,100)
fig, ax = plt.subplots(1,2)

for ai in arange:
    fx = base_expr.subs(a,ai)
    dfx= sym.diff(fx)
    critical_points = sym.solve(dfx)
    #print(critical_points)
    ax[0].plot(xrange, sym.lambdify(x, fx)(xrange))
    ax[1].plot(xrange, sym.lambdify(x, dfx)(xrange))
    
    if 1 in critical_points:
        display(Math('\\Rightarrow %s \\text{ has a critical point at x=1!!!! Wohooo!}'%(sym.latex(fx))))
    elif 2 in critical_points:
        display(Math('\\Rightarrow %s \\text{ has a critical point at x=2!!!! Wohooo!}'%(sym.latex(fx))))
    else:
        display(Math('\\quad  %s \\text{ has NO critical point at x=1 or x=2.}'%(sym.latex(fx))))

        
ax[0].set_ylim([0,2])
ax[0].set_title('Function')
ax[0].plot([1,1],[0,2],'--',color='gray')
ax[0].plot([2,2],[0,2],'--',color='gray')

ax[1].set_ylim([-2,2])
ax[1].set_title('Its derivative')
ax[1].plot([1,1],[-2,2],'--',color='gray')
ax[1].plot([2,2],[-2,2],'--',color='gray')
```
