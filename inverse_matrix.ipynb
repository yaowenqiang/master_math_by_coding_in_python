```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
m = 5
A = np.random.randint(-5,6,(m,m))
# A[:,0] = A[:,1]
```

```python
A_inv = np.linalg.inv(A)
A_inv
```

```python
fig, ax = plt.subplots(1,3)

ax[0].imshow(A)
ax[0].set_title('A')

ax[1].imshow(A_inv)
ax[1].set_title('A$^{-1}$')

ax[2].imshow(A@A_inv)
ax[2].set_title('AA$^{-1}$')
```

```python
m = 5
A = np.random.randint(-m,m+1,(m,m))
A[:,0] = A[:,1]
APinv = np.linalg.pinv(A)
fig, ax = plt.subplots(1,3)

ax[0].imshow(A)
ax[0].set_title('A')

ax[1].imshow(APinv)
ax[1].set_title('A$^*$')

ax[2].imshow(A@APinv)
ax[2].set_title('AA$^*$')

for i in ax:
    i.axis('off')
```

```python
m = 5
A = np.random.randn(m,m+2)
APinv = np.linalg.pinv(A)
fig, ax = plt.subplots(1,3)

ax[0].imshow(A)
ax[0].set_title('A')

ax[1].imshow(APinv)
ax[1].set_title('A$^*$')

ax[2].imshow(A@APinv)
ax[2].set_title('AA$^*$')

for i in ax:
    i.axis('off')
```
