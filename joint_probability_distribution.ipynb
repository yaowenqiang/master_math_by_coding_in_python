```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
import seaborn as sns
```

```python
n = 1000
x = np.random.randn(n)
y = np.random.randn(n) **2

plt.hist2d(x,y,30,vmax=10)
plt.xlabel('x')
plt.ylabel('$x^2$')
```

```python
sns.set()
sns.jointplot(x=x,y=y,kind='scatter',color=[.8,.3, .9])
# sns.jointplot(x=x,y=y,kind='scatter',color=[.8,.3, .9]).plot_joint(sns.kdeplot)

plt.xlabel('x')
plt.ylabel('$x^2$')
plt.show()
```
