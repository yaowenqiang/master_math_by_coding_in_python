```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import pandas as pd
import seaborn as sns
```

```python
max_n = 10000

largest_prim_fact = np.zeros(max_n + 1,dtype=int)/0
for i in range(2,max_n + 1):
    largest_prim_fact[i] = np.max(sym.primefactors(i))
```

```python
largest_prim_fact
```

```python
plt.plot(largest_prim_fact,'k.',markersize=.5)
plt.xlabel('Number')
plt.ylabel('Largest prime factor')
plt.show()
```

```python
unique_max_primes = np.unique(largest_prim_fact)
```

```python
unique_max_primes

```

```python
counts = np.zeros(len(unique_max_primes),dtype=int)
for i,u in enumerate(unique_max_primes):   
    counts[i] = np.sum(largest_prim_fact == u)
    
```

```python
counts
```

```python
plt.plot(unique_max_primes,counts,'k.')
plt.xlim([-2,500])
plt.title('%s is the modal max prime factor' % unique_max_primes[np.argmax(counts)])
plt.show()
```

```python
number_list = np.arange(max_n+1)
number_list[largest_prim_fact<=5]
```

```python
len(number_list[largest_prim_fact<=5])
```

```python
smooth_count = np.zeros(len(unique_max_primes))
for i,u in enumerate(unique_max_primes):
    smooth_count[i] = len(number_list[largest_prim_fact<=u])
    
```

```python
plt.plot(unique_max_primes,smooth_count,'k.')
plt.xlim([-2,500])
plt.xlabel('Smooth number')
plt.ylabel('Number of numbers')
plt.title('K-smooth number frequences')
plt.show()
```
