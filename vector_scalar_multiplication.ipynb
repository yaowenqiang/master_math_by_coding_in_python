```python
import sympy as sym
from IPython.display import display,Math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from mpl_toolkits.mplot3d import Axes3D
import scipy as sp
```

```python
v = np.array([.5,1])
s = [1,-.5,2,.5]
for si in s:
    sv = si * v
    plt.plot([0,sv[0]],[0,sv[1]],'o-',linewidth=3,label='$\\lambda=%g$'%(si))

plt.axis('square')
plt.axis([-3,3,-3,3])
plt.grid()
plt.legend()
```

```python
v1 = np.array([-1,2])
v2 = np.array([1,1])

v3a = v1 + v2
v3b = np.add(v1,v2)
v3c = np.zeros(2)
for i in range(0,2):
    v3c[i] = v1[i] + v2[i]
```

```python
print(v3a)
print(v3b)
print(v3c)
```

```python
plt.plot([0,v1[0]],[0,v1[1]],linewidth=3,label='$v_1$')
plt.plot([0,v2[0]] + v1[0],[0,v2[1]]+ v1[1],linewidth=3,label='$v_2$')
plt.plot([0,v3a[0]],[0,v3a[1]],linewidth=3,label='$v_3$')
s1 = sym.latex(sym.sympify(v1))
s2 = sym.latex(sym.sympify(v2))

s3 = sym.latex(sym.sympify(v3a))

display(Math('%s+%s=%s'%(s1,s2,s3)))
plt.axis('square')

plt.axis([-2,2,-2,4])
plt.legend()
plt.grid()
plt.show()
```
